```python
from datasets import Dataset

ragas_data = {
    "question": [],
    "answer": [],
    "contexts": [],
    "ground_truth": []
}

for item in TEST_SET:

    # question
    if "user_input" in item:
        question = item["user_input"][0]
    else:
        question = item["question"]

    # answer generated by your RAG
    answer = item["response"][0]

    # ground truth reference
    ground_truth = item.get("reference", item.get("references"))[0]

    # retrieved contexts
    contexts = item["retrieved_contexts"][0] if isinstance(item["retrieved_contexts"][0], list) else item["retrieved_contexts"]

    ragas_data["question"].append(question)
    ragas_data["answer"].append(answer)
    ragas_data["contexts"].append(contexts)
    ragas_data["ground_truth"].append(ground_truth)


dataset = Dataset.from_dict(ragas_data)

print(dataset)
```

```python
from ragas import evaluate
from ragas.metrics import (
    faithfulness,
    answer_relevancy,
    context_precision,
    context_recall
)

results = evaluate(
    dataset=dataset,
    metrics=[
        faithfulness,
        answer_relevancy,
        context_precision,
        context_recall
    ],
    llm=ragas_llm,
    embeddings=ragas_embeddings
)

print(results)
```
